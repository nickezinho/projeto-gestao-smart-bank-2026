```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("Planilha dos Investidores.xlsx")

print("Planilha carregada com sucesso!")
print(df.head())
print(df.shape)
print(df.to_string())
```

```python
print("==== INFORMAÇÕES DA BASE ====")
print(df.info())

print("\n==== ESTATÍSTICAS DOS INVESTIDORES ====")
print(df.describe())
```

```python
print("==== MÉDIA DOS INVESTIDORES ====")

print("Idade média:", round(df["IDADE"].mean(), 2))
print("Renda mensal média: R$", round(df["RENDA MENSAL"].mean(), 2))
print("Valor médio investido: R$", round(df["VALOR INVESTIDO"].mean(), 2))
print("Prazo médio:", round(df["PRAZO MESES"].mean(), 2), "meses")
print("Rentabilidade média:", round(df["RENTABILIDADE"].mean() * 100, 2), "%")
print("Valor atual médio: R$", round(df["VALOR ATUAL"].mean(), 2))
```

```python
print("==== TIPOS DE INVESTIMENTO ====")

quantidade_investimentos = df["TIPO DE INVESTIMENTO"].value_counts()

print(quantidade_investimentos)
```

```python
print("==== VALOR INVESTIDO POR TIPO ====")

valor_por_tipo = df.groupby("TIPO DE INVESTIMENTO")["VALOR INVESTIDO"].agg(["count", "sum", "mean"]
)

valor_por_tipo.columns = [
    "Quantidade de Investidores",
    "Valor Total Investido",
    "Investimento Médio"
]


print(valor_por_tipo.sort_values("Valor Total Investido", ascending=False))
```
